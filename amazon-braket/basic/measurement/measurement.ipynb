# Measurement in Different Bases — Amazon Braket

The same quantum state gives different measurement statistics depending
on the basis.  We measure in the Z basis (default) and the X basis
(Hadamard before measurement).

```python
from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
device = LocalSimulator()

def z_basis_measure(prep, shots=1000):
    """Measure in the Z basis (computational basis)."""
    measured = Circuit()
    measured.add_circuit(prep)
    measured.measure(0)
    return device.run(measured, shots=shots).result().result_types[0].value

def x_basis_measure(prep, shots=1000):
    """Measure in the X basis (H before measurement)."""
    measured = Circuit()
    measured.add_circuit(prep)
    measured.h(0)
    measured.measure(0)
    return device.run(measured, shots=shots).result().result_types[0].value

def print_counts(name, counts, shots):
    print(f"  {name}:")
    for state in sorted(counts, key=counts.get, reverse=True):
        frac = counts[state] / shots
        print(f"    |{state}>: {counts[state]:>4}/{shots}  ({frac:.1%})")
    print()
```

## |0\rangle — deterministic in Z, random in X

```python
prep = Circuit()
result = device.run(prep, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")

print_counts("Z-basis", z_basis_measure(prep), 1000)
print_counts("X-basis", x_basis_measure(prep), 1000)
```

## |+\rangle = H|0\rangle — random in Z, deterministic in X

```python
prep = Circuit()
prep.h(0)
result = device.run(prep, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")

print_counts("Z-basis (random 50/50)", z_basis_measure(prep), 1000)
print_counts("X-basis (deterministic |+>)", x_basis_measure(prep), 1000)
```

## |-\rangle = HX|0\rangle — random in Z, deterministic in X

```python
prep = Circuit()
prep.x(0)
prep.h(0)
result = device.run(prep, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")

print_counts("Z-basis (random 50/50)", z_basis_measure(prep), 1000)
print_counts("X-basis (deterministic |->)", x_basis_measure(prep), 1000)
```

## |1\rangle — deterministic in Z, random in X

```python
prep = Circuit()
prep.x(0)
result = device.run(prep, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")

print_counts("Z-basis", z_basis_measure(prep), 1000)
print_counts("X-basis (random 50/50)", x_basis_measure(prep), 1000)
```
