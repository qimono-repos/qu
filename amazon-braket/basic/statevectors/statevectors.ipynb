# Statevector Inspection — Amazon Braket

The statevector describes a quantum state as a linear combination of
computational-basis states.  We create several canonical states and
inspect their amplitudes and probabilities.

```python
import json

from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
device = LocalSimulator()
```

## |+\rangle = H|0\rangle

The $|+\rangle$ state has equal amplitude in $|0\rangle$ and $|1\rangle$:

```python
circuit = Circuit()
circuit.h(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```

## |->\rangle = HX|0\rangle

The $|-\rangle$ state has opposite phases:

```python
circuit = Circuit()
circuit.x(0)
circuit.h(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```

## |+i\rangle = HS|0\rangle

The $|+i\rangle$ state has a complex phase:

```python
circuit = Circuit()
circuit.h(0)
circuit.s(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```

## Tensor product: |+\rangle \u2297 |0\rangle

A two-qubit system where one qubit is in $|+\rangle$ and the other in $|0\rangle$.

```python
circuit = Circuit()
circuit.h(1)  # qubit 1 gets H, qubit 0 stays |0>
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"statevector length: {len(result.result_types[0].value)}")
print(f"probabilities: {result.result_types[1].value}")
```
