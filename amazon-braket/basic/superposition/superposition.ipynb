# Superposition — Amazon Braket

The Hadamard gate creates equal superposition.  We explore single-qubit
superposition and the Bell state, verifying with shot statistics.

```python
import json

from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
device = LocalSimulator()
```

## Single-qubit superposition: H|0\rangle

The Hadamard puts the qubit on the equator of the Bloch sphere.

```python
circuit = Circuit()
circuit.h(0)
circuit.measure(0)
print(circuit)

result = device.run(circuit, shots=4096).result()
print(f"shots=4096: {result.result_types[0].value}")
print("expect roughly half |0> and half |1>")
```

## Bell state |\u03a6+\rangle = (|00\rangle + |11\rangle)/\u221a2

H on qubit 0 followed by CNOT creates an entangled pair.

```python
circuit = Circuit()
circuit.h(0)
circuit.cnot(0, 1)
print(circuit)

# Exact statevector
result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```

```python
# Shot statistics
measured = Circuit()
measured.add_circuit(circuit)
measured.measure(0)
measured.measure(1)

result = device.run(measured, shots=4096).result()
counts = result.result_types[0].value
print(f"|Phi+> shots=4096: {counts}")
print("only |00> and |11> should appear — the qubits are entangled")
```

## Bell state |\u03a8-\rangle (independent circuit)

A different Bell state: $(|01\rangle - |10\rangle)/\sqrt{2}$

```python
circuit = Circuit()
circuit.x(1)
circuit.h(0)
circuit.z(0)
circuit.cnot(0, 1)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```
