# Tensor Products — Amazon Braket

Multi-qubit systems are built by composing single-qubit gates on
separate qubits.  The statevector of a 2-qubit system has 4 entries,
one for each computational basis state.

```python
import json

from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
device = LocalSimulator()
```

## |00\rangle — default two-qubit state

```python
circuit = Circuit()
result = device.run(circuit, shots=0).result()
print(f"probabilities: {result.result_types[1].value}")
```

## |10\rangle — X on qubit 1

```python
circuit = Circuit()
circuit.x(1)
result = device.run(circuit, shots=0).result()
print(f"probabilities: {result.result_types[1].value}")
print("Note: qubit 1 is leftmost in the bitstring, qubit 0 is rightmost.")
```

## |01\rangle — X on qubit 0

```python
circuit = Circuit()
circuit.x(0)
result = device.run(circuit, shots=0).result()
print(f"probabilities: {result.result_types[1].value}")
```

## |11\rangle — X on both qubits

```python
circuit = Circuit()
circuit.x(0)
circuit.x(1)
result = device.run(circuit, shots=0).result()
print(f"probabilities: {result.result_types[1].value}")
```

## |+\rangle \u2297 |0\rangle = (|00\rangle + |10\rangle)/\u221a2

```python
circuit = Circuit()
circuit.h(1)  # H on qubit 1 (left qubit)
result = device.run(circuit, shots=0).result()
print(f"probabilities: {result.result_types[1].value}")
print("The left qubit is in |+>, the right qubit is in |0>.")
```

## |+\rangle \u2297 |+\rangle = (|00\rangle + |01\rangle + |10\rangle + |11\rangle)/2

```python
circuit = Circuit()
circuit.h(0)
circuit.h(1)
result = device.run(circuit, shots=0).result()
print(f"probabilities: {result.result_types[1].value}")
print("All four basis states equally likely — product of two 50/50 states.")
```
