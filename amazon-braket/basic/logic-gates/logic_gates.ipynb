# Basic Logic Gates — Amazon Braket

Each gate is applied to a computational-basis input and simulated
exactly with the local statevector simulator.

```python
import json

from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
device = LocalSimulator()
```

## Pauli-X (NOT)

Flips |0\rangle to |1\rangle.

```python
circuit = Circuit()
circuit.x(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"probabilities: {result.result_types[1].value}")
```

## Pauli-Y

Y|0\rangle = i|1\rangle

```python
circuit = Circuit()
circuit.y(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
```

## Pauli-Z

Z|+\rangle = |-\rangle — flips the phase of |1\rangle.

```python
circuit = Circuit()
circuit.h(0)
circuit.z(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```

## Hadamard

Creates equal superposition from |0\rangle or |1\rangle.

```python
print("H|0>:")
circuit = Circuit()
circuit.h(0)
result = device.run(circuit, shots=0).result()
print(f"  amplitudes: {result.result_types[0].value}")

print("H|1>:")
circuit = Circuit()
circuit.x(0)
circuit.h(0)
result = device.run(circuit, shots=0).result()
print(f"  amplitudes: {result.result_types[0].value}")
```

## S and T Phase Gates

S adds a pi/2 phase to |1\rangle; T adds pi/4.

```python
print("S|+>:")
circuit = Circuit()
circuit.h(0)
circuit.s(0)
result = device.run(circuit, shots=0).result()
print(f"  amplitudes: {result.result_types[0].value}")

print("T|+>:")
circuit = Circuit()
circuit.h(0)
circuit.t(0)
result = device.run(circuit, shots=0).result()
print(f"  amplitudes: {result.result_types[0].value}")
```

## CNOT (CX) Truth Table

CNOT flips the target qubit when the control is |1\rangle.

```python
print(f"  {'input':>8}  ->  {'output':>8}")
print(f"  {'--------':>8}     {'--------':>8}")

for control, target in ((0, 0), (0, 1), (1, 0), (1, 1)):
    circuit = Circuit()
    if target:
        circuit.x(0)
    if control:
        circuit.x(1)
    circuit.cnot(1, 0)  # control=q1, target=q0
    result = device.run(circuit, shots=0).result()
    amps = result.result_types[0].value
    out_idx = next(i for i, a in enumerate(amps) if abs(a) > 1e-10)
    out_bits = format(out_idx, '02b')
    print(f"  |{control}{target}>  ->  |{out_bits}>")
```
