# Computational Basis States — Amazon Braket

The computational basis for a single qubit consists of two states:

- $|0\rangle$ — the default state after qubit initialization
- $|1\rangle$ — obtained by applying a Pauli-X (NOT) gate

We prepare both, inspect their statevectors, and measure them
to show that measurement collapses the state with certainty.

```python
import json

from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
device = LocalSimulator()
```

## |0\rangle — the default qubit state

An empty circuit leaves the qubit in $|0\rangle$.

```python
circuit = Circuit()
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```

## |1\rangle — flip with Pauli-X

```python
circuit = Circuit()
circuit.x(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```

## Superposition for contrast

A Hadamard gate creates an equal superposition — not a basis state.

```python
circuit = Circuit()
circuit.h(0)
print(circuit)

result = device.run(circuit, shots=0).result()
print(f"amplitudes: {result.result_types[0].value}")
print(f"probabilities: {result.result_types[1].value}")
```
