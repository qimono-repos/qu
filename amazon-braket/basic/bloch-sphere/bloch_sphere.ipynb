# Bloch Sphere — Amazon Braket

The Bloch sphere is a geometric representation of single-qubit states.
Each canonical state maps to a point on the sphere:

- |0\rangle: north pole
- |1\rangle: south pole  
- |+\rangle: +x axis
- |-\rangle: -x axis
- |+i\rangle: +y axis

```python
import math
import cmath

from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
device = LocalSimulator()
```

```python
def bloch_angles(alpha, beta):
    """Return (theta, phi) for psi = alpha|0> + beta|1>."""
    theta = 2.0 * math.acos(min(1.0, abs(alpha)))
    if abs(beta) > 1e-12:
        phi = cmath.phase(beta)
    else:
        phi = 0.0
    return theta, phi
```

## |0\rangle — north pole

```python
circuit = Circuit()
result = device.run(circuit, shots=0).result()
amps = result.result_types[0].value
theta, phi = bloch_angles(amps[0], amps[1])
print(f"|0>: alpha={amps[0]}, beta={amps[1]}")
print(f"     theta={theta:.4f} rad ({math.degrees(theta):.1f} deg), phi={phi:.4f} rad ({math.degrees(phi):.1f} deg)")
```

## |1\rangle — south pole

```python
circuit = Circuit()
circuit.x(0)
result = device.run(circuit, shots=0).result()
amps = result.result_types[0].value
theta, phi = bloch_angles(amps[0], amps[1])
print(f"|1>: alpha={amps[0]}, beta={amps[1]}")
print(f"     theta={theta:.4f} rad ({math.degrees(theta):.1f} deg), phi={phi:.4f} rad ({math.degrees(phi):.1f} deg)")
```

## |+\rangle — +x axis

```python
circuit = Circuit()
circuit.h(0)
result = device.run(circuit, shots=0).result()
amps = result.result_types[0].value
theta, phi = bloch_angles(amps[0], amps[1])
print(f"|+>: alpha={amps[0]}, beta={amps[1]}")
print(f"      theta={theta:.4f} rad ({math.degrees(theta):.1f} deg), phi={phi:.4f} rad ({math.degrees(phi):.1f} deg)")
```

## |-\rangle — -x axis

```python
circuit = Circuit()
circuit.x(0)
circuit.h(0)
result = device.run(circuit, shots=0).result()
amps = result.result_types[0].value
theta, phi = bloch_angles(amps[0], amps[1])
print(f"|->: alpha={amps[0]}, beta={amps[1]}")
print(f"      theta={theta:.4f} rad ({math.degrees(theta):.1f} deg), phi={phi:.4f} rad ({math.degrees(phi):.1f} deg)")
```

## |+i\rangle — +y axis

```python
circuit = Circuit()
circuit.h(0)
circuit.s(0)
result = device.run(circuit, shots=0).result()
amps = result.result_types[0].value
theta, phi = bloch_angles(amps[0], amps[1])
print(f"|+i>: alpha={amps[0]}, beta={amps[1]}")
print(f"       theta={theta:.4f} rad ({math.degrees(theta):.1f} deg), phi={phi:.4f} rad ({math.degrees(phi):.1f} deg)")
```
