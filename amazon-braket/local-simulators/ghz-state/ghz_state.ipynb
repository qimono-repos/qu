# GHZ State — Amazon Braket

The GHZ (Greenberger–Horne–Zeilinger) state is a maximally entangled state
of $n$ qubits:

$$|\text{GHZ}\rangle = \frac{1}{\sqrt{2}}(|00\ldots0\rangle + |11\ldots1\rangle)$$

We build it with a Hadamard on qubit 0 followed by a chain of CNOT gates,
then run on the local statevector simulator.

```python
from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
num_qubits = 4

circuit = Circuit()
circuit.h(0)
for i in range(num_qubits - 1):
    circuit.cnot(i, i + 1)
circuit
```

```python
device = LocalSimulator()
task = device.run(circuit, shots=0)
result = task.result()

probabilities = result.result_types[0].value

print("Probabilities:")
for state, prob in sorted(probabilities.items()):
    print(f"  |{state}\u27E9 = {prob:.6f}")
```
