# Bell State — Amazon Braket

A Bell state is the simplest entangled state of two qubits:

$$|\Phi^+\rangle = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$$

We build it with a Hadamard on qubit 0 followed by a CNOT (qubit 0 → qubit 1),
then run on the local statevector simulator.

```python
import json

from braket.circuit import Circuit
from braket.devices import LocalSimulator
```

```python
circuit = Circuit()
circuit.h(0)
circuit.cnot(0, 1)
circuit
```

```python
device = LocalSimulator()
task = device.run(circuit, shots=0)
result = task.result()

state_vector = result.result_types[0].value
probabilities = result.result_types[1].value

print(f"State vector: {state_vector}")
print(f"Probabilities: {json.dumps({k: round(v, 6) for k, v in probabilities.items()})}")
```
