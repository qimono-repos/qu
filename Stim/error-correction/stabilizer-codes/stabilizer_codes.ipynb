# 3-Qubit Stabilizer Codes

Demonstrates the **bit-flip** and **phase-flip** stabilizer codes:
- Encoding logical qubits into physical qubits
- Error injection and syndrome measurement
- Error detection and correction

```python
import stim
import numpy as np
```

## 3-Qubit Bit-Flip Code

Protects against single X (bit-flip) errors:
- Encode: |0⟩ → |000⟩, |1⟩ → |111⟩
- Stabilizers: Z₀Z₁ and Z₁Z₂

```python
circuit = stim.Circuit("""
    # Inject X error on qubit 1
    X 1

    # Syndrome measurement
    DETECTOR rec[-2]
    DETECTOR rec[-1]

    M 0 1 2
""")

sampler = circuit.compile_detector_error_model()
detection_events = sampler.sample(1000)

syndrome_counts = {}
for i in range(1000):
    syndrome = tuple(detection_events[i])
    syndrome_counts[syndrome] = syndrome_counts.get(syndrome, 0) + 1

print("Bit-flip code syndrome distribution:")
for syndrome, count in sorted(syndrome_counts.items()):
    print(f"  {syndrome}: {count}/1000 ({100*count/1000:.1f}%)")
```

## Syndrome Table

| Syndrome (s₀, s₁) | Error location |
|---|---|
| (0, 0) | No error |
| (1, 0) | Qubit 0 |
| (1, 1) | Qubit 1 |
| (0, 1) | Qubit 2 |
