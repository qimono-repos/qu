# Surface Codes (Repetition Code)

Demonstrates the **distance-3 repetition code** using Stim's built-in generator:
- Clean vs noisy circuits
- Detector error model analysis
- Error detection capability

```python
import stim
import numpy as np
```

```python
circuit = stim.Circuit.generated(
    "repetition_code:memory",
    distance=3,
    rounds=1,
    after_clifford_depolarization=0.0,
)

print(f"Qubits: {circuit.num_qubits}")
print(f"Detectors: {circuit.num_detectors}")
print(f"Observables: {circuit.num_observables}")
```
