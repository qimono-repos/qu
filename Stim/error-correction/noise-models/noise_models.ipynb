# Noise Models & Threshold Behavior

Compares **logical vs physical error rates** across different code distances.

```python
import stim
import numpy as np
```

```python
for p in [0.001, 0.01, 0.05]:
    print(f"\np = {p}:")
    for d in [3, 5, 7]:
        circuit = stim.Circuit.generated(
            "repetition_code:memory",
            distance=d,
            rounds=1,
            after_clifford_depolarization=p,
        )
        sampler = circuit.compile_detector_error_model()
        events = sampler.sample(5000)
        rate = np.mean(np.any(events, axis=1))
        print(f"  d={d}: logical error rate = {rate:.4f}")
```
