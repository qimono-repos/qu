# Toffoli Gate (CCNOT)

The Toffoli gate is a 3-qubit gate: it flips the target qubit
if and only if both control qubits are |1>.

It is universal for classical computation.

```python
from pyquil import Program, get_qc
from pyquil.gates import CCNOT, X, MEASURE
```

## Toffoli with |11> input

Set both controls to |1> via X gates. Toffoli flips the target from |0> to |1>.
Expect |111> as the result.

```python
p = Program()
ro = p.declare("ro", "BIT", 3)

p += X(0)
p += X(1)
p += CCNOT(0, 1, 2)
for i in range(3):
    p += MEASURE(i, ro[i])

qc = get_qc("3q-qvm")
results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"Toffoli |110> input -> {counts}")
```

## Toffoli with superposition controls

Put both control qubits in superposition. The Toffoli creates
a 3-qubit entangled state.

```python
from pyquil.gates import H

p = Program()
ro = p.declare("ro", "BIT", 3)

p += H(0)
p += H(1)
p += CCNOT(0, 1, 2)
for i in range(3):
    p += MEASURE(i, ro[i])

results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"Toffoli with superposition -> {counts}")
```
