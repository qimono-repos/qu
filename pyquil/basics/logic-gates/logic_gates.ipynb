# Logic Gates

Single-qubit gates:

| Gate | Matrix | Effect |
|------|--------|--------|
| X | [[0,1],[1,0]] | Bit flip (NOT) |
| Y | [[0,-i],[i,0]] | Bit + phase flip |
| Z | [[1,0],[0,-1]] | Phase flip |
| H | Hadamard | Creates superposition |
| S | [[1,0],[0,i]] | pi/2 phase shift |
| T | [[1,0],[0,e^(i*pi/4)]] | pi/4 phase shift |

```python
from pyquil import Program, get_qc
from pyquil.gates import X, Y, Z, H, S, T, MEASURE
```

## Apply each gate and measure

Starting from |0>, each gate produces a different output.

```python
qc = get_qc("1q-qvm")

gates = [("X", X), ("Y", Y), ("Z", Z), ("H", H), ("S", S), ("T", T)]

for label, gate_fn in gates:
    p = Program()
    ro = p.declare("ro", "BIT", 1)
    p += gate_fn(0)
    p += MEASURE(0, ro[0])
    results = qc.run(qc.compile(p)).readout_data.get("ro")
    counts = {}
    for row in results:
        key = "".join(str(b) for b in row)
        counts[key] = counts.get(key, 0) + 1
    print(f"{label} gate -> Counts: {counts}")
```
