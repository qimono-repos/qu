# Computational Basis

The computational basis states are |0> and |1>.

- |0> is the ground state (all measurements return 0)
- |1> is the excited state (all measurements return 1)

These are the eigenstates of the Pauli-Z operator.

```python
from pyquil import Program, get_qc
from pyquil.gates import X, MEASURE
```

## Prepare |0> and measure

No gate needed — qubits start in |0>.

```python
p = Program()
ro = p.declare("ro", "BIT", 1)
p += MEASURE(0, ro[0])

qc = get_qc("1q-qvm")
results = qc.run(qc.compile(p)).readout_data.get("ro")
print(f"|0> measurements: {results}")
```

## Prepare |1> and measure

Apply X (NOT) gate to flip |0> to |1>.

```python
p = Program()
ro = p.declare("ro", "BIT", 1)
p += X(0)
p += MEASURE(0, ro[0])

results = qc.run(qc.compile(p)).readout_data.get("ro")
print(f"|1> measurements: {results}")
```
