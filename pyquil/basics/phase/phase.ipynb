# Phase Shifts

The PHASE gate applies a phase rotation `e^(i*theta)` to the |1> component:

```
PHASE(theta) = [[1, 0], [0, e^(i*theta)]]
```

Combined with Hadamard gates, phase shifts change the interference pattern.

````python
from pyquil import Program, get_qc
from pyquil.gates import H, PHASE, MEASURE
import numpy as np
````

## Interference with phase shifts

Apply H -> PHASE(theta) -> H. The phase shift changes the
probability of measuring |0> vs |1>.

````python
qc = get_qc("1q-qvm")
thetas = [0.0, np.pi / 4, np.pi / 2, np.pi, 3 * np.pi / 2]

for theta in thetas:
    p = Program()
    ro = p.declare("ro", "BIT", 1)
    p += H(0)
    p += PHASE(theta, 0)
    p += H(0)
    p += MEASURE(0, ro[0])
    results = qc.run(qc.compile(p)).readout_data.get("ro")
    counts = {}
    for row in results:
        key = "".join(str(b) for b in row)
        counts[key] = counts.get(key, 0) + 1
    print(f"theta={theta:.4f} -> Counts: {counts}")
````
