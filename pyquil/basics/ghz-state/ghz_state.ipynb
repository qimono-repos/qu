# GHZ State

Create a 4-qubit GHZ state using a CNOT chain:

1. Apply Hadamard to qubit 0
2. Chain CNOT gates: CNOT(0,1), CNOT(1,2), CNOT(2,3)
3. Measure all qubits — expect `|0000>` and `|1111>` only

```python
from pyquil import Program, get_qc
from pyquil.gates import H, CNOT, MEASURE
```

```python
n = 4
p = Program()
ro = p.declare("ro", "BIT", n)

p += H(0)
for i in range(n - 1):
    p += CNOT(i, i + 1)

for i in range(n):
    p += MEASURE(i, ro[i])

print(p)
```

```python
qc = get_qc(f"{n}q-qvm")
executable = qc.compile(p)
results = qc.run(executable).readout_data.get("ro")

counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1

print(f"Counts: {counts}")
```
