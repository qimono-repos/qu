# Tensor Products

Two-qubit states are formed by tensor products of single-qubit states.

| qubit 0 | qubit 1 | 2-qubit state |
|---------|---------|---------------|
| |0> | |0> | |00> |
| |+> | |0> | (|00> + |10>)/sqrt(2) |

```python
from pyquil import Program, get_qc
from pyquil.gates import H, MEASURE
```

## |+> tensor |0>

Apply H to qubit 0 (superposition), leave qubit 1 in |0>.
Expect only |00> and |10> outcomes.

```python
p = Program()
ro = p.declare("ro", "BIT", 2)

p += H(0)
p += MEASURE(0, ro[0])
p += MEASURE(1, ro[1])

qc = get_qc("2q-qvm")
results = qc.run(qc.compile(p)).readout_data.get("ro")

counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"|+> tensor |0> counts: {counts}")
print("Expect |00> and |10> only (qubit 0 in superposition, qubit 1 in |0>)")
```
