# Superposition

The Hadamard gate puts a qubit into superposition:

```
H|0> = |+> = (|0> + |1>) / sqrt(2)
```

Measuring a superposition gives |0> or |1> randomly with equal probability.

````python
from pyquil import Program, get_qc
from pyquil.gates import H, MEASURE
````

## Create superposition and measure

H|0> produces |+>. With enough shots we see roughly 50/50.

````python
p = Program()
ro = p.declare("ro", "BIT", 1)

p += H(0)
p += MEASURE(0, ro[0])

qc = get_qc("1q-qvm")
results = qc.run(qc.compile(p)).readout_data.get("ro")

counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"Superposition counts: {counts}")
print("H|0> should yield ~50% |0> and ~50% |1>")
````
