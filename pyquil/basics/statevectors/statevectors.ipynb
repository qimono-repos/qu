# Statevectors

The |+> state is created by applying Hadamard to |0>:

```
|+> = (|0> + |1>) / sqrt(2)
```

Measuring |+> gives |0> or |1> with equal probability.

````python
from pyquil import Program, get_qc
from pyquil.gates import H, MEASURE
````

## Prepare |+> and measure

H|0> = |+> = (|0> + |1>) / sqrt(2)

````python
p = Program()
ro = p.declare("ro", "BIT", 1)

p += H(0)
p += MEASURE(0, ro[0])

qc = get_qc("1q-qvm")
results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"|+> measurement counts: {counts}")
print("Expect ~50/50 split between |0> and |1>")
````
