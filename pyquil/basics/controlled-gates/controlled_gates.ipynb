# Controlled Gates

Controlled gates operate on two qubits:

- **CNOT** (CX): Flips target qubit when control is |1>
- **CZ**: Applies Z to target when control is |1>

These are the building blocks of entanglement.

```python
from pyquil import Program, get_qc
from pyquil.gates import CNOT, CZ, H, MEASURE
```

## CNOT demonstration

H on control creates superposition, CNOT entangles the qubits.

```python
p = Program()
ro = p.declare("ro", "BIT", 2)

p += H(0)
p += CNOT(0, 1)
p += MEASURE(0, ro[0])
p += MEASURE(1, ro[1])

qc = get_qc("2q-qvm")
results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"CNOT (H on control) -> {counts}")
```

## CZ demonstration

H on both qubits, then CZ. CZ adds a phase flip when both qubits are |1>.

```python
p = Program()
ro = p.declare("ro", "BIT", 2)

p += H(0)
p += H(1)
p += CZ(0, 1)
p += MEASURE(0, ro[0])
p += MEASURE(1, ro[1])

results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"CZ (H on both) -> {counts}")
```
