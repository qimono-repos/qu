# Measurement in Different Bases

The Z basis is the computational basis {|0>, |1>}.
The X basis is {|+>, |->}.

To measure in the X basis, apply H before measuring in Z.

```python
from pyquil import Program, get_qc
from pyquil.gates import H, MEASURE
```

## Measure |+> in Z basis

|+> = (|0> + |1>)/sqrt(2). Measuring in Z gives ~50/50.

```python
qc = get_qc("1q-qvm")

p = Program()
ro = p.declare("ro", "BIT", 1)
p += H(0)
p += MEASURE(0, ro[0])

results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"|+> in Z basis: {counts}")
```

## Measure |0> in X basis

Apply H-H to |0>: first H creates |+>, second H converts back to |0>.
So H-H on |0> always gives |0>.

```python
p = Program()
ro = p.declare("ro", "BIT", 1)
p += H(0)
p += H(0)
p += MEASURE(0, ro[0])

results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"|0> in X basis (H-H): {counts}")
```
