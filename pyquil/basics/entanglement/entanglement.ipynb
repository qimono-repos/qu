# Entanglement

Entanglement creates correlations between qubits that have no classical
analog. A Bell pair is the simplest entangled state:

```
|Phi+> = (|00> + |11>) / sqrt(2)
```

Measuring one qubit instantly determines the other.

````python
from pyquil import Program, get_qc
from pyquil.gates import H, CNOT, MEASURE
````

## Bell pair

H then CNOT creates the Bell state |Phi+>.

````python
p = Program()
ro = p.declare("ro", "BIT", 2)

p += H(0)
p += CNOT(0, 1)
p += MEASURE(0, ro[0])
p += MEASURE(1, ro[1])

qc = get_qc("2q-qvm")
results = qc.run(qc.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"Bell pair: {counts}")
````

## 3-qubit GHZ state

Extend entanglement to three qubits with a CNOT chain.

````python
p = Program()
ro = p.declare("ro", "BIT", 3)

p += H(0)
p += CNOT(0, 1)
p += CNOT(1, 2)
for i in range(3):
    p += MEASURE(i, ro[i])

qc3 = get_qc("3q-qvm")
results = qc3.run(qc3.compile(p)).readout_data.get("ro")
counts = {}
for row in results:
    key = "".join(str(b) for b in row)
    counts[key] = counts.get(key, 0) + 1
print(f"3-qubit GHZ: {counts}")
````
