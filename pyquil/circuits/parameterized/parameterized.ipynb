# Parameterized Circuits

Quil supports parameterized programs using `Declare` for classical
memory. A `REAL` variable `theta` is bound at execution time via
`memory_map`, avoiding recompilation for each parameter value.

```python
from pyquil import Program, get_qc
from pyquil.gates import RZ, MEASURE
import numpy as np
```

```python
p = Program()
ro = p.declare("ro", "BIT", 1)
theta = p.declare("theta", "REAL")

p += RZ(theta, 0)
p += MEASURE(0, ro[0])

print(p)
```

```python
qc = get_qc("1q-qvm")
executable = qc.compile(p)

thetas = [0.0, np.pi / 4, np.pi / 2, np.pi, 3 * np.pi / 2, 2 * np.pi]

for val in thetas:
    bitstrings = qc.run(executable, memory_map={"theta": [val]})
    ro = bitstrings.readout_data.get("ro")
    print(f"theta={val:.4f} -> {ro}")
```
