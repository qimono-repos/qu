# Statevector inspection

A `Statevector` holds the full quantum state as a complex vector.
This notebook builds several states and inspects their amplitudes,
probabilities, and ket decomposition.

```python
from IPython.display import display
import qiskit as qk
import qiskit_aer as qka
```

```python
def show(qc, title):
    print(title)
    print(qc.draw())
    sv = qk.quantum_info.Statevector.from_instruction(qc)
    for bits, amp in sv.to_dict().items():
        if abs(amp) > 1e-12:
            print(f"  {amp.real:+.3f}{amp.imag:+.3f}j |{bits}>")
    print()


def show_mpl(qc, title):
    print(title)
    display(qc.draw(output="mpl"))
    sv = qk.quantum_info.Statevector.from_instruction(qc)
    for bits, amp in sv.to_dict().items():
        if abs(amp) > 1e-12:
            print(f"  {amp.real:+.3f}{amp.imag:+.3f}j |{bits}>")
    print()
```

## The $|+\rangle$ state

$|+\rangle = H|0\rangle = (|0\rangle + |1\rangle)/\sqrt{2}$.
Equal amplitudes, both positive and real.

```python
qc = qk.QuantumCircuit(1)
qc.h(0)
show_mpl(qc, "|+> = H|0>")

sv = qk.quantum_info.Statevector.from_instruction(qc)
print("raw data:", sv.data)
print("probabilities:", sv.probabilities_dict())
print("norm:", sum(abs(a)**2 for a in sv.data))
```

## The $|-\rangle$ state

$|-\rangle = HX|0\rangle = (|0\rangle - |1\rangle)/\sqrt{2}$.
The $|1\rangle$ amplitude is negative.

```python
qc = qk.QuantumCircuit(1)
qc.x(0)
qc.h(0)
show_mpl(qc, "|-> = HX|0>")

sv = qk.quantum_info.Statevector.from_instruction(qc)
print("probabilities:", sv.probabilities_dict())
```

## The $|+i\rangle$ state

$|+i\rangle = HS|0\rangle = (|0\rangle + i|1\rangle)/\sqrt{2}$.
The $|1\rangle$ amplitude is purely imaginary.

```python
qc = qk.QuantumCircuit(1)
qc.h(0)
qc.s(0)
show_mpl(qc, "|+i> = HS|0>")

sv = qk.quantum_info.Statevector.from_instruction(qc)
print("probabilities:", sv.probabilities_dict())
```

## Two-qubit tensor product

A two-qubit circuit has a 4-element statevector.
$|+\rangle \otimes |0\rangle = (|00\rangle + |10\rangle)/\sqrt{2}$.

```python
qc = qk.QuantumCircuit(2)
qc.h(1)
show(qc, "|+> x |0>")
display(qc.draw(output="mpl"))
sv = qk.quantum_info.Statevector.from_instruction(qc)
print("probabilities:", sv.probabilities_dict())
```

## Summary

- `Statevector` gives the exact amplitudes before measurement.
- `probabilities_dict()` gives $|\alpha|^2$ for each basis state.
- Global phase (overall complex factor) does not affect probabilities.
