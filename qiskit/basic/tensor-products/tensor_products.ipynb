# Tensor products

When two qubits are independent, the combined state is their
**tensor product**: $|a\rangle \otimes |b\rangle$.
Qiskit uses **little-endian** bitstrings: qubit 0 is the rightmost bit,
qubit 1 is the leftmost.  So $|01\rangle$ means qubit 1 = 0, qubit 0 = 1.

```python
from IPython.display import display
import qiskit as qk
import qiskit_aer as qka
```

```python
def show(qc, title):
    print(title)
    print(qc.draw())
    sv = qk.quantum_info.Statevector.from_instruction(qc)
    for bits, amp in sv.to_dict().items():
        if abs(amp) > 1e-12:
            print(f"  {amp.real:+.3f}{amp.imag:+.3f}j |{bits}>")
    print()


def show_mpl(qc, title):
    print(title)
    display(qc.draw(output="mpl"))
    sv = qk.quantum_info.Statevector.from_instruction(qc)
    for bits, amp in sv.to_dict().items():
        if abs(amp) > 1e-12:
            print(f"  {amp.real:+.3f}{amp.imag:+.3f}j |{bits}>")
    print()
```

## $|00\rangle$: default two-qubit state

Two fresh qubits give $|00\rangle$. The statevector has 4 elements
but only one is nonzero.

```python
qc = qk.QuantumCircuit(2)
show_mpl(qc, "|00> (default)")
sv = qk.quantum_info.Statevector.from_instruction(qc)
print("probabilities:", sv.probabilities_dict())
```

## $|10\rangle$: $X$ on qubit 1

The left qubit (qubit 1) is flipped. In the bitstring $|10\rangle$
the left digit is qubit 1.

```python
qc = qk.QuantumCircuit(2)
qc.x(1)
show_mpl(qc, "|10> (X on qubit 1)")
```

## $|01\rangle$: $X$ on qubit 0

The right qubit (qubit 0) is flipped.

```python
qc = qk.QuantumCircuit(2)
qc.x(0)
show_mpl(qc, "|01> (X on qubit 0)")
```

## $|11\rangle$: both flipped

```python
qc = qk.QuantumCircuit(2)
qc.x(0)
qc.x(1)
show_mpl(qc, "|11> (X on both)")
```

## $|+\rangle \otimes |0\rangle$

Apply $H$ to qubit 1 only: the left qubit is in superposition,
the right qubit stays $|0\rangle$.

```python
qc = qk.QuantumCircuit(2)
qc.h(1)
show_mpl(qc, "|+> x |0> = (|00> + |10>)/sqrt(2)")
sv = qk.quantum_info.Statevector.from_instruction(qc)
print("probabilities:", sv.probabilities_dict())
```

## $|+\rangle \otimes |+\rangle$

Both qubits in superposition: all four basis states equally likely.

```python
qc = qk.QuantumCircuit(2)
qc.h(0)
qc.h(1)
show_mpl(qc, "|+> x |+> = (|00> + |01> + |10> + |11>)/2")
sv = qk.quantum_info.Statevector.from_instruction(qc)
print("probabilities:", sv.probabilities_dict())
```

## Summary

- Two independent qubits: $|a\rangle \otimes |b\rangle$ is a tensor
  product with 4 amplitudes.
- Qiskit bitstrings: qubit 0 = rightmost, qubit 1 = leftmost.
- Applying a gate to one qubit does not affect the other in a
  product state.
