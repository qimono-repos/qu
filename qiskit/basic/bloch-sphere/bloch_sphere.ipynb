# Bloch sphere visualization

The Bloch sphere maps every single-qubit state to a point on
a unit sphere.  $|0\rangle$ sits at the north pole, $|1\rangle$
at the south pole, and equal superpositions on the equator.

Qiskit's `plot_bloch_multivector` draws the statevector on this sphere.

```python
from IPython.display import display
import numpy as np
import qiskit as qk
from qiskit.visualization import plot_bloch_multivector
```

```python
def show_mpl(qc, title):
    print(title)
    display(qc.draw(output="mpl"))
    sv = qk.quantum_info.Statevector.from_instruction(qc)
    for bits, amp in sv.to_dict().items():
        if abs(amp) > 1e-12:
            print(f"  {amp.real:+.3f}{amp.imag:+.3f}j |{bits}>")
    display(plot_bloch_multivector(sv, title=title))
    print()
```

## $|0\rangle$: north pole

The qubit arrow points straight up.

```python
qc = qk.QuantumCircuit(1)
show_mpl(qc, "|0> — north pole")
```

## $|1\rangle$: south pole

The $X$ gate flips the arrow from north to south.

```python
qc = qk.QuantumCircuit(1)
qc.x(0)
show_mpl(qc, "|1> = X|0> — south pole")
```

## $|+\rangle$: equator ($+x$ direction)

$H|0\rangle$ puts the arrow on the equator pointing along $+x$.

```python
qc = qk.QuantumCircuit(1)
qc.h(0)
show_mpl(qc, "|+> — equator, +x")
```

## $|-\rangle$: equator ($-x$ direction)

$HX|0\rangle$ flips to the opposite side of the equator.

```python
qc = qk.QuantumCircuit(1)
qc.x(0)
qc.h(0)
show_mpl(qc, "|-> = HX|0> — equator, -x")
```

## $|+i\rangle$: equator ($+y$ direction)

$HS|0\rangle$ adds a $\pi/2$ phase, rotating the arrow to $+y$.

```python
qc = qk.QuantumCircuit(1)
qc.h(0)
qc.s(0)
show_mpl(qc, "+i> = HS|0> — equator, +y")
```

## Summary

| State | Bloch position |
|-------|----------------|
| $|0\rangle$ | north pole $(0, 0, 1)$ |
| $|1\rangle$ | south pole $(0, 0, -1)$ |
| $|+\rangle$ | equator $+x$ $(1, 0, 0)$ |
| $|-\rangle$ | equator $-x$ $(-1, 0, 0)$ |
| $|+i\rangle$ | equator $+y$ $(0, 1, 0)$ |

Any single-qubit gate moves the arrow on this sphere.
