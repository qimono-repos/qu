# Computational-basis states

The two computational-basis states are $|0\rangle$ and $|1\rangle$.
Every qubit starts in $|0\rangle$. The $X$ gate flips it to $|1\rangle$.
Measuring a basis state always returns that state with certainty.

```python
from IPython.display import display
import qiskit as qk
import qiskit_aer as qka
```

```python
def show(qc, title):
    print(title)
    print(qc.draw())
    sv = qk.quantum_info.Statevector.from_instruction(qc)
    for bits, amp in sv.to_dict().items():
        if abs(amp) > 1e-12:
            print(f"  {amp.real:+.3f}{amp.imag:+.3f}j |{bits}>")
    print()


def run_shots(qc, shots=2000):
    measured = qc.copy()
    if measured.num_clbits == 0:
        measured.measure_all()
    backend = qka.AerSimulator()
    compiled = qk.transpile(measured, backend)
    return backend.run(compiled, shots=shots).result().get_counts()
```

## $|0\rangle$: the default qubit state

A fresh `QuantumCircuit(1)` holds $|0\rangle$. No gate needed.

```python
qc = qk.QuantumCircuit(1)
show(qc, "|0> state")
display(qc.draw(output="mpl"))
print("shots:", run_shots(qc))
```

## $|1\rangle$: flip with the $X$ gate

$X|0\rangle = |1\rangle$. Measuring always returns 1.

```python
qc = qk.QuantumCircuit(1)
qc.x(0)
show(qc, "|1> = X|0>")
display(qc.draw(output="mpl"))
print("shots:", run_shots(qc))
```

## For comparison: equal superposition

$H|0\rangle = (|0\rangle + |1\rangle)/\sqrt{2}$. Measuring gives
random 50/50 results.

```python
qc = qk.QuantumCircuit(1)
qc.h(0)
show(qc, "|+> = H|0>")
display(qc.draw(output="mpl"))
counts = run_shots(qc)
print("shots:", counts)
display(qk.visualization.plot_histogram(counts, title="|+> measurement"))
```

## Summary

| State | Circuit | Measurement result |
|-------|---------|--------------------|
| $|0\rangle$ | empty | always 0 |
| $|1\rangle$ | $X$ | always 1 |
| $|+\rangle$ | $H$ | 50/50 random |

Basis states are deterministic; superpositions are not.
