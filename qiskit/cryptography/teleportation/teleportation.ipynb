# Quantum Teleportation

Alice teleports an arbitrary state $|\psi\rangle$ to Bob using a Bell
pair, a Bell measurement, and classical corrections.

```python
import qiskit as qk
import qiskit_aer as qka
```

## Prepare state and teleport

```python
theta, phi = 0.7, 1.2

psi = qk.QuantumCircuit(1)
psi.ry(theta, 0)
psi.rz(phi, 0)
sv = qk.quantum_info.Statevector.from_instruction(psi)
print(f"State to teleport: {sv}")
```

```python
qc = qk.QuantumCircuit(3, 3)
qc.ry(theta, 0)
qc.rz(phi, 0)
qc.h(1)
qc.cx(1, 2)
qc.cx(0, 1)
qc.h(0)
qc.measure(0, 0)
qc.measure(1, 1)
qc.x(2).c_if(1, 1)
qc.z(2).c_if(0, 1)
qc.measure(2, 2)
print(qc.draw())
```

```python
sim = qka.AerSimulator()
compiled = qk.transpile(qc, sim)
counts = sim.run(compiled, shots=4096).result().get_counts()
total = sum(counts.values())
m0 = sum(v for k, v in counts.items() if k[0] == '0')
m1 = sum(v for k, v in counts.items() if k[0] == '1')
print(f"Target qubit: |0>={m0}/{total} ({m0/total:.3f}), |1>={m1}/{total} ({m1/total:.3f})")
probs = sv.probabilities_dict()
print(f"Expected:     |0> ~ {probs.get('0', 0):.3f}, |1> ~ {probs.get('1', 0):.3f}")
```
