# Quantum Random Number Generation

Measure Hadamard-created superpositions to extract truly random bits.
Statistical tests verify the randomness.

```python
import qiskit as qk
import qiskit_aer as qka
```

## Single random bit

```python
def quantum_random_bit():
    qc = qk.QuantumCircuit(1, 1)
    qc.h(0)
    qc.measure(0, 0)
    backend = qka.AerSimulator()
    compiled = qk.transpile(qc, backend)
    counts = backend.run(compiled, shots=1).result().get_counts()
    return int(list(counts.keys())[0], 2)

bits = [quantum_random_bit() for _ in range(16)]
print(f"16 random bits: {''.join(map(str, bits))}")
```

## Batch generation with chi-squared test

```python
def generate_random_bits(n):
    qc = qk.QuantumCircuit(n, n)
    for i in range(n):
        qc.h(i)
    qc.measure(range(n), range(n))
    backend = qka.AerSimulator()
    compiled = qk.transpile(qc, backend)
    counts = backend.run(compiled, shots=1).result().get_counts()
    return [int(b) for b in list(counts.keys())[0].zfill(n)]


import math

bits = generate_random_bits(1000)
ones = sum(bits)
zeros = len(bits) - ones
expected = len(bits) / 2
chi2 = ((zeros - expected)**2 + (ones - expected)**2) / expected
print(f"1000 bits: 0s={zeros}, 1s={ones}")
print(f"Chi-squared: {chi2:.4f} (critical value 3.841 at p=0.05)")
print(f"{'PASS' if chi2 < 3.841 else 'FAIL'}")
```
