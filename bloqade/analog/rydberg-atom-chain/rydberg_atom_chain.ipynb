# Rydberg Atom Chain — Bloqade (Analog)

A 1D chain of neutral atoms, each a two-level system with ground
|g⟩ and Rydberg |r⟩ states. The Rydberg blockade mechanism prevents
neighboring atoms from both being excited when the inter-atomic
spacing is within the blockade radius Rb.

We define 4 atoms in a line, apply a Rabi frequency pulse to
excite them, then sweep the detuning to map out the phase diagram.

```python
import numpy as np

from bloqade.analog import AtomArrangement, Sequence
from bloqade.analog.emulator import Emulator
```

```python
n_atoms = 4
spacing_um = 6.0

atoms = AtomArrangement()
for i in range(n_atoms):
    atoms.add(x=i * spacing_um, y=0.0)

atoms
```

```python
seq = Sequence(atoms)

seq.set.uniform.rabi.frequency.value = 2 * np.pi * 1.0
seq.set.uniform.detuning.value = 2 * np.pi * 0.0

seq.set.uniform.rabi.frequency.value = 0.0
seq.set.uniform.detuning.value = 2 * np.pi * -3.0

seq.measure.basis = "ground-rydberg"
seq
```

```python
device = Emulator()
job = device.run(seq)
result = job.report()

print(f"Bitstrings: {result.counts}")
print(f"Mean rydberg population: {result.mean()}")
```
