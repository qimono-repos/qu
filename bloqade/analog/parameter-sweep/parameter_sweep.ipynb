# Parameter Sweep — Bloqade (Analog)

By sweeping the inter-atomic spacing while keeping the laser
parameters fixed, we can observe the transition between the
**ordered phase** (strong blockade, atoms cannot both be excited)
and the **disordered phase** (weak interaction, atoms behave
independently).

The blockade radius Rb ≈ 7 µm for Rydberg atoms. When spacing
< Rb, the blockade suppresses excitations. When spacing > Rb,
each atom can be excited independently.

```python
import numpy as np

from bloqade.analog import AtomArrangement, Sequence
from bloqade.analog.emulator import Emulator
```

```python
def build_sequence(spacing_um: float, detuning_mhz: float) -> Sequence:
    n_atoms = 4
    atoms = AtomArrangement()
    for i in range(n_atoms):
        atoms.add(x=i * spacing_um, y=0.0)

    seq = Sequence(atoms)
    seq.set.uniform.rabi.frequency.value = 2 * np.pi * 1.0
    seq.set.uniform.detuning.value = 2 * np.pi * 0.0
    seq.set.uniform.rabi.frequency.value = 0.0
    seq.set.uniform.detuning.value = 2 * np.pi * detuning_mhz
    seq.measure.basis = "ground-rydberg"
    return seq
```

```python
spacings = [4.0, 6.0, 8.0, 10.0]
detuning_mhz = -1.0
device = Emulator()

print(f"Detuning: 2π × {detuning_mhz} MHz (constant)")
for spacing in spacings:
    seq = build_sequence(spacing, detuning_mhz)
    job = device.run(seq)
    result = job.report()
    print(f"  spacing = {spacing:5.1f} µm → mean Rydberg pop = {result.mean():.4f}")
```

Small spacing: blockade suppresses excitations (ordered phase).  
Large spacing: atoms act independently (disordered phase).
