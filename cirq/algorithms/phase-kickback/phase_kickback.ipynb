# Phase Kickback

Phase kickback transfers the eigenvalue of a unitary to the control qubit.
This is the key mechanism behind many quantum algorithms.

```python
import cirq
import numpy as np
```

## CNOT Phase Kickback

Control in |+\u27e9, target in |1\u27e9: CNOT flips target, but since |1\u27e9 is an eigenstate
of X, the -1 phase kicks back to the control.

```python
q0, q1 = cirq.LineQubit.range(2)
sim = cirq.Simulator()

circuit = cirq.Circuit(cirq.H(q0), cirq.X(q1), cirq.CNOT(q0, q1))
print(circuit)
sv = sim.simulate(circuit).final_state_vector
print(f"\nState: {np.array2string(sv, precision=3)}")
print("Result: (|0\u27e9|1\u27e9 - |1\u27e9|0\u27e9)/\u221a2")
```

## CZ Phase Kickback

Both qubits in |+\u27e9: CZ flips the phase of |11\u27e9, creating an entangled state.

```python
circuit2 = cirq.Circuit(cirq.H(q0), cirq.H(q1), cirq.CZ(q0, q1))
print(circuit2)
sv2 = sim.simulate(circuit2).final_state_vector
print(f"\nState: {np.array2string(sv2, precision=3)}")
```

## Eigenvalue Kickback (T gate)

Target in eigenstate |1\u27e9 of T gate. The T eigenvalue kicks back to control.

```python
circuit_t = cirq.Circuit(
    cirq.H(q0), cirq.X(q1),
    cirq.CNOT(q0, q1), cirq.T(q1), cirq.CNOT(q0, q1),
)
print(circuit_t)
sv_t = sim.simulate(circuit_t).final_state_vector
print(f"\nState: {np.array2string(sv_t, precision=3)}")
print("Control picks up T eigenvalue when target is |1\u27e9.")
```
