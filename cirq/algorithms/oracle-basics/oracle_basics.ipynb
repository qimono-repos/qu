# Oracle Basics

A quantum oracle is a black-box operation that marks specific input states.
The simplest example: an oracle that flips the phase of |11⟩ using a CZ gate.

```python
import cirq
import numpy as np
```

## Phase Oracle for |11⟩

CZ(q0, q1) applies -1 phase to |11⟩ and leaves |00⟩, |01⟩, |10⟩ unchanged.

```python
q0, q1 = cirq.LineQubit.range(2)
oracle = cirq.Circuit(cirq.CZ(q0, q1))
print(oracle)

sim = cirq.Simulator()
print("\nTruth table:")
for a in range(2):
    for b in range(2):
        ops = []
        if a: ops.append(cirq.X(q0))
        if b: ops.append(cirq.X(q1))
        ops.append(cirq.CZ(q0, q1))
        sv = sim.simulate(cirq.Circuit(ops)).final_state_vector
        print(f"  |{a}{b}\u27e9 -> {np.array2string(sv, precision=3)}")
```

## Oracle on Superposition

Apply H to both qubits to create equal superposition, then apply the oracle.

```python
circuit = cirq.Circuit(cirq.H(q0), cirq.H(q1), cirq.CZ(q0, q1))
sv = sim.simulate(circuit).final_state_vector
print("State after oracle on |++\u27e9:")
for i in range(4):
    print(f"  |{i:02b}\u27e9: amplitude={sv[i]:.4f}, prob={abs(sv[i])**2:.4f}")
```

## Oracle in a Search Circuit

H-H-Oracle-H-H amplifies the marked state.

```python
search = cirq.Circuit(
    cirq.H(q0), cirq.H(q1),
    cirq.CZ(q0, q1),
    cirq.H(q0), cirq.H(q1),
)
print(search)
result = sim.simulate(search)
sv = result.final_state_vector
probs = np.abs(sv) ** 2
for i in range(4):
    print(f"  P(|{i:02b}\u27e9) = {probs[i]:.4f}")
```
