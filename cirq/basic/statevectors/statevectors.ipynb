# State Vectors

A qubit's state is described by a 2D complex vector (state vector).
By inspecting the state vector directly, we can see the amplitudes
of each computational basis state.

```python
import numpy as np
import cirq
```

## |+⟩ State

Apply Hadamard to |0⟩ to get the equal superposition |+⟩ = (|0⟩ + |1⟩)/√2.

```python
q = cirq.LineQubit(0)
circuit = cirq.Circuit(cirq.H(q))
print(circuit)

result = cirq.Simulator().simulate(circuit)
sv = result.final_state_vector
print(f"\nState vector: {np.array2string(sv, precision=4)}")
print(f"Amplitude of |0⟩: {sv[0]:.4f}")
print(f"Amplitude of |1⟩: {sv[1]:.4f}")
print(f"Probabilities: {np.abs(sv)**2}")
```

## Multiple Single-Qubit States

Let's inspect several common single-qubit states.

```python
states = {
    "|0⟩": [],
    "|1⟩": [cirq.X(q)],
    "|+⟩": [cirq.H(q)],
    "|-⟩": [cirq.X(q), cirq.H(q)],
    "|i⟩": [cirq.H(q), cirq.S(q)],
}

for label, ops in states.items():
    result = cirq.Simulator().simulate(cirq.Circuit(ops))
    sv = result.final_state_vector
    print(f"{label}: {np.array2string(sv, precision=4)}  |  |amp|² = {np.abs(sv)**2}")
```
