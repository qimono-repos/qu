# Tensor Products

Multi-qubit states are formed by taking the tensor (Kronecker) product
of individual qubit states. For two qubits:
|a⟩ ⊗ |b⟩ = |ab⟩ (a 4-element vector).

```python
import numpy as np
import cirq
```

## Computational Basis Tensor Products

```python
q0, q1 = cirq.LineQubit.range(2)
sim = cirq.Simulator()

states = {
    "|00⟩": [],
    "|01⟩": [cirq.X(q1)],
    "|10⟩": [cirq.X(q0)],
    "|11⟩": [cirq.X(q0), cirq.X(q1)],
}

for label, ops in states.items():
    sv = sim.simulate(cirq.Circuit(ops)).final_state_vector
    print(f"{label}: {sv}")
```

## Superposition Tensor Products

|+⟩ ⊗ |0⟩ = (|00⟩ + |10⟩)/√2

```python
sv = sim.simulate(cirq.Circuit(cirq.H(q0))).final_state_vector
print(f"|+⟩ ⊗ |0⟩ = {np.array2string(sv, precision=4)}")

sv = sim.simulate(cirq.Circuit(cirq.H(q0), cirq.H(q1))).final_state_vector
print(f"|+⟩ ⊗ |+⟩ = {np.array2string(sv, precision=4)}")
```

## Manual Kronecker Product

```python
a = np.array([1/np.sqrt(2), 1/np.sqrt(2)])  # |+⟩
b = np.array([0, 1])                          # |1⟩
print(f"|+⟩ ⊗ |1⟩ (numpy) = {np.kron(a, b)}")

sv = sim.simulate(cirq.Circuit(cirq.H(q0), cirq.X(q1))).final_state_vector
print(f"|+⟩ ⊗ |1⟩ (Cirq)  = {np.array2string(sv, precision=4)}")
```
