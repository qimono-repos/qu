# Bloch Sphere

Any single-qubit pure state can be mapped to a point on the Bloch sphere:
|ψ⟩ = cos(θ/2)|0⟩ + e^{iφ} sin(θ/2)|1⟩.

The coordinates (x, y, z) are the expectation values of the Pauli operators.

```python
import numpy as np
import cirq
```

```python
def bloch_components(state_vector):
    """Extract Bloch sphere (x, y, z) from a single-qubit state vector."""
    rho = np.outer(state_vector, np.conj(state_vector))
    x = 2 * rho[0, 1].real
    y = 2 * rho[0, 1].imag
    z = rho[0, 0].real - rho[1, 1].real
    return x, y, z
```

## Cardinal Points on the Bloch Sphere

```python
q = cirq.LineQubit(0)
sim = cirq.Simulator()

states = {
    "|0⟩ (north)": [],
    "|1⟩ (south)": [cirq.X(q)],
    "|+⟩ (east)": [cirq.H(q)],
    "|-⟩ (west)": [cirq.X(q), cirq.H(q)],
    "|+i⟩ (front)": [cirq.H(q), cirq.S(q)],
    "|-i⟩ (back)": [cirq.X(q), cirq.H(q), cirq.S(q)],
}

print(f"{'State':<20} {'x':>7} {'y':>7} {'z':>7}")
print("-" * 44)
for label, ops in states.items():
    sv = sim.simulate(cirq.Circuit(ops)).final_state_vector
    x, y, z = bloch_components(sv)
    print(f"{label:<20} {x:7.4f} {y:7.4f} {z:7.4f}")
```

## Rotations Around Each Axis

```python
for name, rot in [("X", cirq.rx), ("Y", cirq.ry), ("Z", cirq.rz)]:
    print(f"\n{name}-axis rotations of |0⟩:")
    for angle in [np.pi/4, np.pi/2, np.pi]:
        sv = sim.simulate(cirq.Circuit(rot(angle)(q))).final_state_vector
        x, y, z = bloch_components(sv)
        print(f"  θ={angle:.2f} → ({x:.3f}, {y:.3f}, {z:.3f})")
```
