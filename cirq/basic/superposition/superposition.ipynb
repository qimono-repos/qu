# Superposition

A qubit in superposition exists in both |0⟩ and |1⟩ simultaneously.
The Hadamard gate creates equal superposition from |0⟩.

```python
import cirq
```

## Single-Qubit Superposition

H|0⟩ = (|0⟩ + |1⟩)/√2. Measurement yields 0 or 1 with equal probability.

```python
q = cirq.LineQubit(0)
circuit = cirq.Circuit(cirq.H(q), cirq.measure(q, key="m"))
print(circuit)

result = cirq.Simulator().run(circuit, repetitions=1000)
counts = result.histogram(key="m")
print(f"\nCounts (1000 shots): {counts}")
print(f"P(|0⟩) ≈ {counts[0]/1000:.3f},  P(|1⟩) ≈ {counts[1]/1000:.3f}")
```

## State Vector Before Measurement

```python
result = cirq.Simulator().simulate(cirq.Circuit(cirq.H(q)))
sv = result.final_state_vector
print(f"|ψ⟩ = {sv[0]:.4f}|0⟩ + {sv[1]:.4f}|1⟩")
print(f"Probabilities: |0⟩={abs(sv[0])**2:.4f}, |1⟩={abs(sv[1])**2:.4f}")
```

## Bell Pair: Two-Qubit Superposition

H on qubit 0 + CNOT creates the Bell state (|00⟩ + |11⟩)/√2.

```python
q0, q1 = cirq.LineQubit.range(2)
bell = cirq.Circuit(cirq.H(q0), cirq.CNOT(q0, q1), cirq.measure(q0, q1, key="m"))
print(bell)

result = cirq.Simulator().run(bell, repetitions=1000)
print(f"Bell pair counts: {result.histogram(key='m')}")
```
