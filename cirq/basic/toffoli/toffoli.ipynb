# Toffoli (CCX) Gate

The Toffoli gate is a 3-qubit gate: two controls and one target.
It flips the target qubit only when **both** control qubits are |1⟩.

CCX|abc⟩ = |abc ⊕ (a ∧ b)⟩

```python
import numpy as np
import cirq
```

```python
q0, q1, q2 = cirq.LineQubit.range(3)
print(f"Circuit: {cirq.Circuit(cirq.CCX(q0, q1, q2))}")
print(f"\nMatrix:\n{np.array2string(np.array(cirq.CCX.unitary()), precision=0)}")
```

## Truth Table

The target (q2) flips only when both controls are |1⟩.

```python
sim = cirq.Simulator()

print(f"{'Input':<10} {'Output':<10}")
print("-" * 20)

for a in range(2):
    for b in range(2):
        for c in range(2):
            ops = []
            if a: ops.append(cirq.X(q0))
            if b: ops.append(cirq.X(q1))
            if c: ops.append(cirq.X(q2))
            ops.append(cirq.CCX(q0, q1, q2))
            sv = sim.simulate(cirq.Circuit(ops)).final_state_vector
            out = np.argmax(np.abs(sv))
            print(f"|{a}{b}{c}⟩      |{out:03b}⟩")
```

## Gate Decomposition

The Toffoli can be decomposed into H, T, T†, and CNOT gates.
This is important because most hardware only supports 1- and 2-qubit gates.

```python
decomp = cirq.Circuit(
    cirq.H(q2), cirq.CNOT(q1, q2), cirq.T(q2)**-1, cirq.CNOT(q0, q2),
    cirq.T(q2), cirq.CNOT(q1, q2), cirq.T(q2)**-1, cirq.CNOT(q0, q2),
    cirq.T(q1), cirq.T(q2), cirq.H(q2),
    cirq.CNOT(q0, q1), cirq.T(q0), cirq.T(q1)**-1, cirq.CNOT(q0, q1),
)
print(f"Decomposed circuit ({len(decomp)} operations):")
print(decomp)
```
