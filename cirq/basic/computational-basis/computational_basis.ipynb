# Computational Basis States

The computational basis for a single qubit consists of two states:
- **|0⟩** — the default state of a qubit
- **|1⟩** — obtained by applying a bit-flip (X gate)

Measurement in the computational (Z) basis collapses the qubit to one of these two states.

```python
import cirq
```

## |0⟩ State

A freshly initialized qubit is in the |0⟩ state. Measuring it always yields 0.

```python
q = cirq.LineQubit(0)
circuit_zero = cirq.Circuit(cirq.measure(q, key="m"))
print(circuit_zero)

result = cirq.Simulator().run(circuit_zero, repetitions=10)
print("Counts:", result.histogram(key="m"))
```

## |1⟩ State

Apply an X gate (bit-flip) to flip |0⟩ → |1⟩, then measure.

```python
circuit_one = cirq.Circuit(cirq.X(q), cirq.measure(q, key="m"))
print(circuit_one)

result = cirq.Simulator().run(circuit_one, repetitions=10)
print("Counts:", result.histogram(key="m"))
```

## Summary

| State | Gate | Measurement result |
|-------|------|--------------------|
| |0⟩   | —    | Always 0           |
| |1⟩   | X    | Always 1           |
