# Controlled Gates

Controlled gates act on two qubits: a **control** and a **target**.
The target qubit's gate is applied only when the control qubit is |1⟩.

```python
import numpy as np
import cirq
```

## CNOT (Controlled-NOT)

Flips the target qubit when the control is |1⟩.

```python
q0, q1 = cirq.LineQubit.range(2)
print(f"CNOT circuit: {cirq.Circuit(cirq.CNOT(q0, q1))}")
print(f"CNOT matrix:\n{np.array2string(np.array(cirq.CNOT.unitary()), precision=0)}")
```

```python
sim = cirq.Simulator()

print("CNOT truth table:")
for a, b in [(0,0), (0,1), (1,0), (1,1)]:
    ops = []
    if a: ops.append(cirq.X(q0))
    if b: ops.append(cirq.X(q1))
    ops.append(cirq.CNOT(q0, q1))
    sv = sim.simulate(cirq.Circuit(ops)).final_state_vector
    out = np.argmax(np.abs(sv))
    print(f"  |{a}{b}⟩ → |{out:02b}⟩  (state: {np.array2string(sv, precision=3)})")
```

## CZ (Controlled-Z)

Applies a Z gate to the target when control is |1⟩.
Equivalently: flips the phase of |11⟩.

```python
print(f"CZ circuit: {cirq.Circuit(cirq.CZ(q0, q1))}")
print(f"CZ matrix:\n{np.array2string(np.array(cirq.CZ.unitary()), precision=0)}")

print("\nCZ truth table:")
for a, b in [(0,0), (0,1), (1,0), (1,1)]:
    ops = []
    if a: ops.append(cirq.X(q0))
    if b: ops.append(cirq.X(q1))
    ops.append(cirq.CZ(q0, q1))
    sv = sim.simulate(cirq.Circuit(ops)).final_state_vector
    out = np.argmax(np.abs(sv))
    print(f"  |{a}{b}⟩ → |{out:02b}⟩  (state: {np.array2string(sv, precision=3)})")
```

## Reversed Control

CNOT(q1, q0) — control is q1, target is q0.

```python
print(cirq.Circuit(cirq.CNOT(q1, q0)))
```
