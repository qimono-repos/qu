# Single-Qubit Logic Gates

Cirq provides standard single-qubit gates. Each gate is a unitary matrix
that transforms the qubit's state vector.

```python
import numpy as np
import cirq
```

## Gate Matrix and Effect on |0⟩

```python
q = cirq.LineQubit(0)
sim = cirq.Simulator()

gates = [
    ("X (bit-flip)", cirq.X),
    ("Y", cirq.Y),
    ("Z (phase-flip)", cirq.Z),
    ("H (Hadamard)", cirq.H),
    ("S (√Z)", cirq.S),
    ("T (√S)", cirq.T),
]

for name, gate in gates:
    result = sim.simulate(cirq.Circuit(gate(q)))
    sv = result.final_state_vector
    print(f"{name}:")
    print(f"  Matrix:\n{np.array2string(np.array(gate.unitary()), precision=3)}")
    print(f"  |0⟩ → {np.array2string(sv, precision=4)}\n")
```

## Circuit Diagrams

Each gate applied to a single qubit.

```python
for name, gate in gates:
    print(f"{name}: {cirq.Circuit(gate(q))}")
```
