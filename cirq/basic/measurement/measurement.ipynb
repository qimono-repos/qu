# Measurement Bases

The choice of measurement basis affects the statistics:
- **Z basis** (computational): measure directly
- **X basis**: apply H, then measure in Z

Same state, different basis → different results.

```python
import numpy as np
import cirq
```

## Z-Basis Measurement

```python
q = cirq.LineQubit(0)
sim = cirq.Simulator()

for label, ops in [("H|0⟩", [cirq.H(q)]), ("H|1⟩", [cirq.X(q), cirq.H(q)])]:
    circuit = cirq.Circuit(ops + [cirq.measure(q, key="m")])
    result = sim.run(circuit, repetitions=1000)
    print(f"{label} → Z-meas: {result.histogram(key='m')}")
```

## X-Basis Measurement

To measure in the X basis, apply H before measuring.
|+⟩ → eigenvalue +1 → result 0.
|−⟩ → eigenvalue -1 → result 1.

```python
for label, ops in [("|+⟩", [cirq.H(q)]), ("|-⟩", [cirq.X(q), cirq.H(q)])]:
    circuit = cirq.Circuit(ops + [cirq.H(q), cirq.measure(q, key="m")])
    result = sim.run(circuit, repetitions=1000)
    print(f"{label} → X-meas: {result.histogram(key='m')}")
```

## Same State, Different Bases

Ry(π/4)|0⟩ gives different results when measured in Z vs X.

```python
state_ops = [cirq.ry(np.pi / 4)(q)]

z_circuit = cirq.Circuit(state_ops + [cirq.measure(q, key="m")])
x_circuit = cirq.Circuit(state_ops + [cirq.H(q), cirq.measure(q, key="m")])

z_result = sim.run(z_circuit, repetitions=1000)
x_result = sim.run(x_circuit, repetitions=1000)

print(f"Z-basis: {z_result.histogram(key='m')}")
print(f"X-basis: {x_result.histogram(key='m')}")
```
