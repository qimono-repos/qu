# Phase in Quantum Computing

Phase is a subtle but crucial concept:
- **Global phase** — unobservable, does not affect measurements
- **Relative phase** — affects interference, observable in superpositions

```python
import numpy as np
import cirq
```

## Global Phase (Unobservable)

Multiplying all amplitudes by e^{iθ} changes the state vector
but has no effect on measurement.

```python
sim = cirq.Simulator()

for theta in [0, np.pi / 4, np.pi / 2]:
    circuit = cirq.Circuit(cirq.GlobalPhaseGate(theta))
    sv = sim.simulate(circuit).final_state_vector
    print(f"θ={theta:.2f}: {sv}  (|0⟩ probability = {abs(sv[0])**2:.1f})")
```

## Relative Phase

The S gate applies a π/2 phase to the |1⟩ component.
This changes the state vector but not the Z-basis measurement statistics.

```python
q = cirq.LineQubit(0)

pairs = [
    ("S|+⟩", [cirq.H(q), cirq.S(q)]),
    ("S|-⟩", [cirq.X(q), cirq.H(q), cirq.S(q)]),
]

for label, ops in pairs:
    sv = sim.simulate(cirq.Circuit(ops)).final_state_vector
    print(f"{label}: {np.array2string(sv, precision=4)}")

print("\nMeasurement is the same (50/50) for both:")
for label, ops in pairs:
    circuit = cirq.Circuit(ops + [cirq.measure(q, key="m")])
    result = sim.run(circuit, repetitions=100)
    print(f"  {label}: {result.histogram(key='m')}")
```
