# Oracle Basics

A quantum oracle is a black-box operation that marks certain states by flipping their phase. The simplest example is a **CZ gate** that flips the phase of |11⟩.

```python
import pennylane as qml
```

## Oracle marking |11⟩

The CZ gate applies a phase of −1 to |11⟩ and leaves all other states unchanged.

```python
dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def oracle_circuit():
    qml.PauliX(wires=0)
    qml.PauliX(wires=1)
    qml.CZ(wires=[0, 1])
    return qml.state()

state = oracle_circuit()
print("Oracle applied to |11⟩:")
print(qml.draw(oracle_circuit)())
print(f"State: {state}")
print(f"|11⟩ amplitude: {state[3]:.4f}  (phase flipped to -1)")
```

## CZ truth table

Applying CZ to each computational basis state confirms only |11⟩ is affected.

```python
dev_all = qml.device("default.qubit", wires=2)

@qml.qnode(dev_all)
def apply_cz(state):
    for i, bit in enumerate(state):
        if bit == "1":
            qml.PauliX(wires=i)
    qml.CZ(wires=[0, 1])
    return qml.state()

print("CZ truth table:")
for bits in ["00", "01", "10", "11"]:
    s = apply_cz(bits)
    target = int(bits, 2)
    amp = s[target]
    sign = "-" if amp.real < 0 else "+"
    print(f"  CZ|{bits}\u27e9 = {sign}|{bits}\u27e9   amp={amp:.1f}")
```
