# Phase Kickback

When a controlled-U gate is applied with the target in an eigenstate |u⟩ of U, the eigenvalue phase is **kicked back** into the control qubit's amplitude. This is the foundation of quantum phase estimation.

```python
import pennylane as qml
import numpy as np
```

## Phase kickback with |−⟩ ancilla

Prepare the ancilla in |−⟩ = (|0⟩ − |1⟩)/√2 before the controlled gate. The phase eigenvalue appears on the ancilla.

```python
dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def phase_kickback_circuit(theta):
    qml.Hadamard(wires=0)
    qml.PauliX(wires=1)
    qml.Hadamard(wires=1)
    qml.ctrl(qml.RZ, control=0)(theta, wires=1)
    qml.Hadamard(wires=0)
    return qml.state()

for theta in [np.pi / 4, np.pi / 2, np.pi]:
    print(f"\u03b8 = {theta:.4f}  ({theta / np.pi:.4f}\u03c0)")
    print(qml.draw(phase_kickback_circuit)(theta))
    state = phase_kickback_circuit(theta)
    print(f"  State: {state}")
    print(f"  |00\u27e9 amp = {state[0]:.4f},  |01\u27e9 amp = {state[1]:.4f}")
    print(f"  |10\u27e9 amp = {state[2]:.4f},  |11\u27e9 amp = {state[3]:.4f}")
```

## Without phase kickback (|0⟩ ancilla)

If the ancilla starts in |0⟩ instead of |−⟩, the phase is not visible — it stays internal.

```python
@qml.qnode(dev)
def compare_no_kickback(theta):
    qml.Hadamard(wires=0)
    qml.ctrl(qml.RZ, control=0)(theta, wires=1)
    qml.Hadamard(wires=0)
    return qml.state()

theta = np.pi / 2
state0 = compare_no_kickback(theta)
print(f"Without |\u2212\u27e9 ancilla (\u03b8 = {theta:.4f}):")
print(f"  State: {state0}")
print(f"  P(ancilla=0) = {abs(state0[0])**2 + abs(state0[1])**2:.4f}")
print(f"  P(ancilla=1) = {abs(state0[2])**2 + abs(state0[3])**2:.4f}")
```
