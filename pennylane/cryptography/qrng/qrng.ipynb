# Quantum Random Number Generation (PennyLane)

Measure Hadamard superpositions to extract truly random bits.
Chi-squared test verifies uniformity.

```python
import pennylane as qml
import numpy as np
```

```python
dev = qml.device("default.qubit", wires=1, shots=1)

@qml.qnode(dev)
def random_bit():
    qml.Hadamard(wires=0)
    return qml.sample(wires=0)

bits = [int(random_bit()) for _ in range(16)]
print(f"16 random bits: {''.join(map(str, bits))}")
```

```python
all_bits = [int(random_bit()) for _ in range(1000)]
ones = sum(all_bits)
zeros = len(all_bits) - ones
expected = len(all_bits) / 2
chi2 = ((zeros - expected)**2 + (ones - expected)**2) / expected
print(f"1000 bits: 0s={zeros}, 1s={ones}")
print(f"Chi-squared: {chi2:.4f} (critical value 3.841)")
print(f"{'PASS' if chi2 < 3.841 else 'FAIL'}")
```
