# Quantum Teleportation (PennyLane)

Teleport an arbitrary state using a Bell pair and classical corrections.

```python
import pennylane as qml
import numpy as np

dev = qml.device("default.qubit", wires=3)
```

## Teleportation circuit

```python
@qml.qnode(dev)
def teleport_circuit(theta, phi):
    qml.RY(theta, wires=0)
    qml.RZ(phi, wires=0)
    qml.Hadamard(wires=1)
    qml.CNOT(wires=[1, 2])
    qml.CNOT(wires=[0, 1])
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[1, 2])
    qml.CZ(wires=[0, 2])
    return qml.probs(wires=2)


@qml.qnode(dev)
def original_state(theta, phi):
    qml.RY(theta, wires=0)
    qml.RZ(phi, wires=0)
    return qml.probs(wires=0)

print(qml.draw(teleport_circuit)(0.7, 1.2))
```

```python
for name, theta, phi in [("|0>", 0, 0), ("|1>", np.pi, 0), ("|+>", np.pi/2, 0), ("arb", 0.7, 1.2)]:
    orig = original_state(theta, phi)
    tele = teleport_circuit(theta, phi)
    print(f"{name}: orig={orig[0]:.3f},{orig[1]:.3f}  tele={tele[0]:.3f},{tele[1]:.3f}  match={np.allclose(orig, tele, atol=0.05)}")
```
