# Bell state via PennyLane

Create a Bell pair |Φ+⟩ = (|00⟩+|11⟩)/√2 using PennyLane.
H on qubit 0, then CNOT onto qubit 1.

```python
import pennylane as qml
```

## Build and draw the circuit

```python
N_QUBITS = 2
dev = qml.device("default.qubit", wires=N_QUBITS)

@qml.qnode(dev)
def bell_state():
    qml.Hadamard(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.probs(wires=range(N_QUBITS))

print(qml.draw(bell_state)())
```

## Probabilities

```python
probs = bell_state()
labels = [f"|{i:0{N_QUBITS}b}⟩" for i in range(2**N_QUBITS)]
for label, p in zip(labels, probs):
    print(f"  {label}  {p:.4f}")
```
