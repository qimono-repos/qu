# Computational Basis States

The computational basis states $|0\rangle$ and $|1\rangle$ are the fundamental building blocks of quantum computing. This notebook demonstrates creating these states and measuring their probabilities.

```python
import pennylane as qml
import numpy as np
```

## Define a device and circuits

```python
dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def circuit_zero():
    return qml.probs(wires=0)

@qml.qnode(dev)
def circuit_one():
    qml.PauliX(wires=0)
    return qml.probs(wires=0)
```

## |0⟩ and |1⟩ probabilities

```python
p0 = circuit_zero()
p1 = circuit_one()

print(f"|0>  P(|0>) = {p0[0]:.4f},  P(|1>) = {p0[1]:.4f}")
print(f"|1>  P(|0>) = {p1[0]:.4f},  P(|1>) = {p1[1]:.4f}")
```

## Measurement samples

With finite shots, the statistical results converge to the probabilities above.

```python
dev_shots = qml.device("default.qubit", wires=1, shots=1000)

@qml.qnode(dev_shots)
def sample_zero():
    return qml.sample(wires=0)

@qml.qnode(dev_shots)
def sample_one():
    qml.PauliX(wires=0)
    return qml.sample(wires=0)

s0 = sample_zero()
s1 = sample_one()

print(f"|0> samples:  {np.sum(s0 == 0)} zeros, {np.sum(s0 == 1)} ones")
print(f"|1> samples:  {np.sum(s1 == 0)} zeros, {np.sum(s1 == 1)} ones")
print()
print("The |0> state always measures 0; the |1> state always measures 1.")
```
