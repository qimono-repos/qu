# Measurement

Measurement is how we extract classical information from quantum states. PennyLane provides `qml.expval()` for expectation values, `qml.probs()` for probabilities, and `qml.sample()` for individual shots.

```python
import pennylane as qml
import numpy as np
```

## Expectation values

The expectation value $\langle O \rangle = \langle \psi | O | \psi \rangle$ gives the average outcome of many measurements of observable $O$.

```python
dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def ez_one():
    qml.PauliX(wires=0)
    return qml.expval(qml.Z(0))

@qml.qnode(dev)
def ex_plus():
    qml.Hadamard(wires=0)
    return qml.expval(qml.X(0))

@qml.qnode(dev)
def ez_zero():
    return qml.expval(qml.Z(0))

print(f"<0|Z|0> = {ez_zero():+.4f}  (expected +1)")
print(f"<1|Z|1> = {ez_one():+.4f}  (expected -1)")
print(f"<+|X|+> = {ex_plus():+.4f}  (expected +1)")
```

## Probabilities

```python
@qml.qnode(dev)
def probs_state():
    qml.Hadamard(wires=0)
    return qml.probs(wires=0)

probs = probs_state()
print(f"H|0>  P(|0>) = {probs[0]:.4f},  P(|1>) = {probs[1]:.4f}")
```

## Measurement in different bases

Measuring $|0\rangle$ in the X-basis (via Hadamard) gives $\langle X \rangle = 0$.

```python
@qml.qnode(dev)
def measure_z_basis():
    qml.Hadamard(wires=0)
    qml.PauliX(wires=0)
    return qml.expval(qml.Z(0))

@qml.qnode(dev)
def measure_x_basis():
    qml.Hadamard(wires=0)
    return qml.expval(qml.X(0))

print(f"Measuring X|0> in Z-basis: <Z> = {measure_z_basis():+.4f}  (expected 0)")
print(f"Measuring H|0> in X-basis: <X> = {measure_x_basis():+.4f}  (expected 1)")
```

## Sampling

```python
dev_shots = qml.device("default.qubit", wires=1, shots=4000)

@qml.qnode(dev_shots)
def sample_hadamard():
    qml.Hadamard(wires=0)
    return qml.sample(wires=0)

samples = sample_hadamard()
print(f"H|0> measured 4000 times: {int(np.sum(samples == 0))} zeros, "
      f"{int(np.sum(samples == 1))} ones (expect ~50/50)")
```
