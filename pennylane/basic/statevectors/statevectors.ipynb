# State Vectors

Quantum states are described by **state vectors** — complex-valued vectors in Hilbert space. This notebook demonstrates several single-qubit states and inspects their state vectors via `qml.state()`.

```python
import pennylane as qml
import numpy as np
```

## Define circuits for common states

```python
dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def state_zero():
    return qml.state()

@qml.qnode(dev)
def state_one():
    qml.PauliX(wires=0)
    return qml.state()

@qml.qnode(dev)
def state_plus():
    qml.Hadamard(wires=0)
    return qml.state()

@qml.qnode(dev)
def state_minus():
    qml.PauliX(wires=0)
    qml.Hadamard(wires=0)
    return qml.state()

@qml.qnode(dev)
def state_i():
    qml.Hadamard(wires=0)
    qml.S(wires=0)
    return qml.state()
```

## Inspect state vectors

```python
for label, fn in [
    ("|0>", state_zero),
    ("|1>", state_one),
    ("|+> = H|0>", state_plus),
    ("|-> = XH|0>", state_minus),
    ("|i> = SH|0>", state_i),
]:
    sv = fn()
    print(f"{label}:  {sv}")
```

## Verify normalization

Every valid quantum state has $\|\psi\| = 1$.

```python
for label, fn in [("|0>", state_zero), ("|+>", state_plus), ("|i>", state_i)]:
    sv = fn()
    norm = np.linalg.norm(sv)
    print(f"  ||{label}|| = {norm:.6f}")

print("All state vectors are normalized (norm = 1).")
```
