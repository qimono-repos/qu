# Quantum Logic Gates

Quantum logic gates are the building blocks of quantum circuits, analogous to classical logic gates. Each gate is a unitary matrix that transforms the qubit state.

```python
import pennylane as qml
import numpy as np
```

## Define gates

```python
dev = qml.device("default.qubit", wires=1)

@qml.qnode(dev)
def apply_x():
    qml.PauliX(wires=0)
    return qml.state()

@qml.qnode(dev)
def apply_y():
    qml.PauliY(wires=0)
    return qml.state()

@qml.qnode(dev)
def apply_z():
    qml.PauliZ(wires=0)
    return qml.state()

@qml.qnode(dev)
def apply_h():
    qml.Hadamard(wires=0)
    return qml.state()

@qml.qnode(dev)
def apply_s():
    qml.S(wires=0)
    return qml.state()

@qml.qnode(dev)
def apply_t():
    qml.T(wires=0)
    return qml.state()
```

## Apply each gate to |0⟩

```python
gates = [
    ("PauliX (NOT)", apply_x),
    ("PauliY", apply_y),
    ("PauliZ (phase flip)", apply_z),
    ("Hadamard", apply_h),
    ("S (phase)", apply_s),
    ("T (π/8)", apply_t),
]

for label, fn in gates:
    sv = fn()
    probs = np.abs(sv) ** 2
    print(f"{label}:")
    print(f"  state: {sv}")
    print(f"  probs: P(|0>) = {probs[0]:.4f},  P(|1>) = {probs[1]:.4f}")
    print()
```

## Summary

- **PauliX**: flips |0> → |1> (quantum NOT)
- **PauliY**: combines X and Z rotations
- **PauliZ**: flips the phase of |1> but leaves |0> unchanged
- **Hadamard**: creates equal superposition from |0>
- **S and T**: phase gates that rotate the global phase of |1>
