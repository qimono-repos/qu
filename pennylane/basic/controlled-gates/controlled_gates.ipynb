# Controlled Gates

Controlled gates apply a unitary to a **target** qubit only when the **control** qubit is in the $|1\rangle$ state. The most common is CNOT (controlled-NOT).

```python
import pennylane as qml
import numpy as np
```

## CNOT gate

CNOT flips the target qubit when the control qubit is $|1\rangle$.

```python
dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def cnot_01():
    qml.PauliX(wires=0)
    qml.CNOT(wires=[0, 1])
    return qml.state()

@qml.qnode(dev)
def cnot_10():
    qml.PauliX(wires=1)
    qml.CNOT(wires=[1, 0])
    return qml.state()

@qml.qnode(dev)
def flip_flop():
    qml.CNOT(wires=[0, 1])
    return qml.state()

print("CNOT(0,1) on |10>:", cnot_01(), "-> |11>")
print("CNOT(1,0) on |01>:", cnot_10(), "-> |11>")
print("CNOT(0,1) on |00>:", flip_flop(), "-> |00> (no flip)")
```

## CZ gate

CZ applies a phase flip to $|11\rangle$ only.

```python
@qml.qnode(dev)
def cz_gate():
    qml.Hadamard(wires=0)
    qml.Hadamard(wires=1)
    qml.CZ(wires=[0, 1])
    return qml.state()

sv = cz_gate()
probs = np.abs(sv) ** 2
print(f"H|0> x H|0>, then CZ:")
print(f"  state: {sv}")
print(f"  probs: [{', '.join(f'{p:.4f}' for p in probs)}]")
print(f"  CZ flips phase of |11> component.")
```

## CRZ gate

CRZ applies $RZ(\theta)$ to the target when the control is $|1\rangle$.

```python
@qml.qnode(dev)
def crz_gate():
    qml.PauliX(wires=0)
    qml.Hadamard(wires=1)
    qml.CRZ(np.pi / 4, wires=[0, 1])
    return qml.state()

sv = crz_gate()
probs = np.abs(sv) ** 2
print(f"X q0, H q1, CRZ(pi/4):")
print(f"  state: {sv}")
print(f"  probs: [{', '.join(f'{p:.4f}' for p in probs)}]")
print(f"  CRZ applies RZ to target when control is |1>.")
```
