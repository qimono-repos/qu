# Tensor Products

Multi-qubit states are constructed via the **tensor product** (Kronecker product) of individual qubit states. A two-qubit system lives in a 4-dimensional Hilbert space.

```python
import pennylane as qml
import numpy as np
```

## Define circuits for two-qubit states

Basis order: $|00\rangle$, $|01\rangle$, $|10\rangle$, $|11\rangle$

```python
dev = qml.device("default.qubit", wires=2)

@qml.qnode(dev)
def tensor_zero_zero():
    return qml.state()

@qml.qnode(dev)
def tensor_zero_one():
    qml.PauliX(wires=1)
    return qml.state()

@qml.qnode(dev)
def tensor_one_zero():
    qml.PauliX(wires=0)
    return qml.state()

@qml.qnode(dev)
def tensor_plus_plus():
    qml.Hadamard(wires=0)
    qml.Hadamard(wires=1)
    return qml.state()

@qml.qnode(dev)
def tensor_plus_zero():
    qml.Hadamard(wires=0)
    return qml.state()
```

## Inspect two-qubit states

```python
states = [
    ("|0>|0>", tensor_zero_zero),
    ("|0>|1>", tensor_zero_one),
    ("|1>|0>", tensor_one_zero),
    ("|+>|+>", tensor_plus_plus),
    ("|+>|0>", tensor_plus_zero),
]

for label, fn in states:
    sv = fn()
    probs = np.abs(sv) ** 2
    print(f"{label}:")
    print(f"  state: {sv}")
    print(f"  probs: [{', '.join(f'{p:.4f}' for p in probs)}]")
    print()
```

## Verify against Kronecker product

The circuit output should match the mathematical tensor product.

```python
h_state = np.array([1 / np.sqrt(2), 1 / np.sqrt(2)])
z0 = np.array([1, 0])

kron_result = np.kron(h_state, z0)
sv = tensor_plus_zero()

print(f"|+>|0> via kron:     {kron_result}")
print(f"|+>|0> via circuit:  {sv}")
print(f"Match: {np.allclose(kron_result, sv)}")
```
