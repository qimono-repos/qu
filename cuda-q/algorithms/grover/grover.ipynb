# Grover's Search Algorithm

Grover's algorithm finds a marked item in an unsorted database of $N$ items using only $O(\sqrt{N})$ queries, a quadratic speedup over classical $O(N)$.

The algorithm alternates between:
1. **Oracle**: flips the phase of the marked state
2. **Diffusion operator**: amplifies the amplitude of the marked state

For $N = 8$ (3 qubits), the optimal number of iterations is approximately $\frac{\pi}{4}\sqrt{N} \approx 2$.

```python
import cudaq
import numpy as np


@cudaq.kernel
def grover_101():
    """Grover search for |101> in 3 qubits with 1 iteration."""
    qubits = cudaq.qvector(3)
    h(qubits[0])
    h(qubits[1])
    h(qubits[2])
    x(qubits[0])
    x(qubits[2])
    cz(qubits[0], qubits[2])
    x(qubits[0])
    x(qubits[2])
    h(qubits[0])
    h(qubits[1])
    h(qubits[2])
    x(qubits[0])
    x(qubits[1])
    x(qubits[2])
    cz(qubits[0], qubits[2])
    x(qubits[0])
    x(qubits[1])
    x(qubits[2])
    h(qubits[0])
    h(qubits[1])
    h(qubits[2])
```

```python
basis = ["|000>", "|001>", "|010>", "|011>",
         "|100>", "|101>", "|110>", "|111>"]

print("=== Searching for |101> with 1 iteration ===")
sv = np.array(cudaq.get_state(grover_101))
probs = np.abs(sv) ** 2
for b, amp, prob in zip(basis, sv, probs):
    print(f"  {b}: amp={amp:+.4f}  P={prob:.4f}")
result = cudaq.sample(grover_101, shots_count=1000)
for bitstring, count in result.items():
    print(f"  measured |{bitstring}>: {count}")
```

```python
N = 8
optimal = int(np.round(np.pi / 4 * np.sqrt(N)))
print(f"N = {N}, optimal iterations = {optimal}")
print(f"Classical: O(N) = O({N}) queries")
print(f"Quantum:   O(sqrt(N)) = O({int(np.sqrt(N))}) queries")
```
