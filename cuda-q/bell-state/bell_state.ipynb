# Bell State

The Bell state $|\Phi^+\rangle = \frac{1}{\sqrt{2}}(|00\rangle + |11\rangle)$ is the simplest entangled two-qubit state. We build it with CUDA-Q's `@cudaq.kernel` decorator and sample with `cudaq.sample()`.

```python
import cudaq


@cudaq.kernel
def bell():
    qubits = cudaq.qvector(2)
    h(qubits[0])
    cx(qubits[0], qubits[1])


result = cudaq.sample(bell, shots_count=1000)
print("Bell state results:")
for bitstring, count in result.items():
    print(f"  |{bitstring}>: {count}")
```
